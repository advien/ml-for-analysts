
# Text Classification — Категория товара по описанию

**Цель:** показать базовую текстовую классификацию.  
**Данные:** короткие описания товаров → категории (`clothes`, `electronics`, `books`).  
**Модель:** `CountVectorizer` + `LogisticRegression`.  
**Метрики:** accuracy, confusion matrix.

> Самодостаточный ноут с синтетикой.


```python
# Импорты
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

np.random.seed(42)
print("Импорты готовы.")
```


## 1. Синтетические текстовые данные


```python
clothes = [
    "cotton t-shirt with print", "denim jeans slim fit", "warm winter jacket",
    "summer dress floral pattern", "sports socks pack", "hoodie with zipper",
    "leather belt black", "running shoes breathable", "wool scarf striped",
    "linen shirt short sleeve", "cap adjustable", "raincoat waterproof"
]

electronics = [
    "wireless headphones bluetooth", "smartphone with dual camera",
    "laptop 15 inch ssd", "gaming mouse dpi adjustable", "tv 4k hdr",
    "usb c fast charger", "smartwatch heart rate", "tablet with stylus",
    "portable speaker loud", "external hard drive", "webcam full hd",
    "action camera waterproof"
]

books = [
    "novel historical fiction", "data science handbook",
    "machine learning for beginners", "cookbook healthy recipes",
    "travel guide europe", "fantasy epic saga", "business strategy case studies",
    "self help productivity", "children picture book", "poetry collection",
    "thriller mystery crime", "biography inspiring leader"
]

texts = clothes + electronics + books
labels = (["clothes"] * len(clothes)) + (["electronics"] * len(electronics)) + (["books"] * len(books))

df = pd.DataFrame({"text": texts, "label": labels})
df.sample(5, random_state=1)
```


## 2. Разбиение и векторизация


```python
X_train, X_test, y_train, y_test = train_test_split(
    df["text"], df["label"], test_size=0.3, random_state=42, stratify=df["label"]
)

vectorizer = CountVectorizer()
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

X_train_vec.shape, X_test_vec.shape
```


## 3. Обучение и оценка


```python
clf = LogisticRegression(max_iter=1000)
clf.fit(X_train_vec, y_train)

y_pred = clf.predict(X_test_vec)

acc = accuracy_score(y_test, y_pred)
print("Accuracy:", round(acc, 3))
print("\nClassification report:\n", classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred, labels=["clothes","electronics","books"])
print("Confusion matrix:\n", cm)
```


## 4. Самые важные слова (по коэффициентам логистической регрессии)


```python
import numpy as np
feature_names = np.array(vectorizer.get_feature_names_out())
classes = clf.classes_

topn = 8
for i, cls in enumerate(classes):
    coefs = clf.coef_[i]
    top_idx = np.argsort(coefs)[-topn:]
    print(f"Top слова для класса '{cls}':", feature_names[top_idx])
```
